```python
numeric_columns = df.select_dtypes(include='number').columns

outlier_summary = []

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        'Column': col,
        'Q1': Q1,
        'Q3': Q3,
        'Lower Limit': lower,
        'Upper Limit': upper,
        'Outlier Count': count
    })

outlier_summary_df = pd.DataFrame(outlier_summary)

outlier_summary_df
```

```python
#CORELATION HEATMAP
#How to interpret it:

#Values near +1 indicate a strong positive linear relationship.

#Values near -1 indicate a strong negative linear relationship.

#Values near 0 indicate a weak linear relationship.

numeric_df = df.select_dtypes(include="number")

plt.figure(figsize=(11, 7))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numeric Variables")
plt.tight_layout()
plt.show()
```

```python
#COUNT OF CUSTOMERS IN EACH REGION

region_counts = df["region"].value_counts()
plt.figure(figsize=(8, 8))

plt.pie(
    region_counts,
    labels=region_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    shadow=False
)

plt.title("Netflix Customers by Region")
plt.axis("equal")
plt.show()
```

```python
# Overall churn rate
overall_churn = df["churned"].mean() * 100

print(f"Overall Customer Churn Rate: {overall_churn:.2f}%")
```

```python
#Churn Rate by region

region_summary = (
    df.groupby("region", dropna=False)
    .agg(
        Total_Customers=("churned", "size"),
        Churned_Customers=("churned", "sum"),
        Churn_Rate=("churned", "mean")
    )
    .reset_index()
)

region_summary["Churn_Rate"] *= 100
region_summary = region_summary.sort_values(
    "Churn_Rate", ascending=False
)

plt.figure(figsize=(10, 5))

ax = sns.barplot(
    data=region_summary,
    x="region",
    y="Churn_Rate",
    color="steelblue"
)

ax.bar_label(ax.containers[0], fmt="%.1f%%", padding=3)

plt.axhline(
    overall_churn,
    color="red",
    linestyle="--",
    label=f"Overall Churn: {overall_churn:.1f}%"
)

plt.title("Customer Churn Rate by Region")
plt.xlabel("Region")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=25, ha="right")
plt.legend()
plt.tight_layout()
plt.show()

display(region_summary.round(2))
```

```python
#Churn Rate by device

device_summary = (
    df.groupby("device", dropna=False)
    .agg(
        Total_Customers=("churned", "size"),
        Churned_Customers=("churned", "sum"),
        Churn_Rate=("churned", "mean")
    )
    .reset_index()
)

device_summary["Churn_Rate"] *= 100
device_summary = device_summary.sort_values(
    "Churn_Rate", ascending=False
)

plt.figure(figsize=(10, 5))

ax = sns.barplot(
    data=device_summary,
    x="device",
    y="Churn_Rate",
    color="steelblue"
)

ax.bar_label(ax.containers[0], fmt="%.1f%%", padding=3)

plt.axhline(
    overall_churn,
    color="red",
    linestyle="--",
    label=f"Overall Churn: {overall_churn:.1f}%"
)

plt.title("Customer Churn Rate by Device")
plt.xlabel("Device")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=25, ha="right")
plt.legend()
plt.tight_layout()
plt.show()

display(device_summary.round(2))
```
